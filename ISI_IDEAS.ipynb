```python
# Function

# factorial of a number: 5!
def factorial(number):
  if(number<0):
    print("Factorial does not exist for negative numbers")
  elif(number==0):
    return 1
  else:
    answer = 1
    for i in range(1,number+1):
      answer = answer*i
    return answer

print("552! =",factorial(552))
```

```python
# 1! + 2! + 3! + 4! + ...

num = int(input("Enter the number: ",))
sum = 0
for i in range(1,num+1):
  sum = sum + factorial(i)
  print(sum)
```

```python
# print all the prime numbers in the range lb, ub:

lb = int(input("Enter the lower bound: "))
ub = int(input("Enter the upper bound: "))

def isPrime(x):
  for i in range(2,x):
    if x%i == 0:
      return False
  return True

for i in range(lb,ub+1):
  if isPrime(i):
    print(i, end=" ")


# x=7
# y=8
# print(isPrime(x))
# print(isPrime(y))
```

```python
# Amstrong Number: 153 = 1*3 + 5*3 + 3*3

def isArmstrong(num):
    variables = list(num)
    total = 0
    for i in range(len(num)):
        total += pow(int(variables[i]),len(num))
    return int(num) == total

lb = int(input("Enter the lower bound: "))
ub = int(input("Enter the upper bound: "))

for i in range(lb,ub+1):
  if isArmstrong(str(i)):
    print(i, end=" ")

# num = input("Enter the number: ")
# print(isArmstrong(num))
```

```python
# Tuples : Immutable datatype

# Multiple item in single variable
# Anything decleared is unchangeable

myTuple = (1,2,3,4,5,5,5)
print(myTuple)
print(myTuple[0])

# Ctypecasting a tuple into a list incase of necessary updatation
myTuple = list(myTuple)
print(myTuple)
myTuple[0] = 100
print(myTuple)
myTuple.append(6)
print(myTuple)
myTuple = tuple(myTuple)
print(myTuple)
```
